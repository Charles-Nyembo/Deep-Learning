# Dérivée et descente du gradient

```python
import matplotlib.pyplot as plt 
import numpy as np
```

# Compréhension intuitive de la dérivée
# $f(x)= 2x^2-3x+4$

```python
def f(x):
    return 2*x**2-3*x+4
f(3)
```

Utilisons matplotlib pour tracer la fonction

```python
xs=np.arange(-5,5,0.25)
xy=f(xs)
plt.plot(xs,xy)
```

La dérivée donne la pente de la tangente à la courbe en un point. Pour la calculer, on utilise la formule : $f'(x) = \lim_{h\to0} \frac{f(x+h)-f(x)}{h}$ Avec un petit $h$, on peut estimer la pente numériquement.

**Note :** La pente montre comment $y$ change quand $x$ varie. Si $x$ augmente de 1 et $y$ de 2, la pente est 2.


```python
h= 0.0001
x= -1.0
print("Dérivée en x=-1: ",(f(x+h)-f(x))/h)
x= 2
print("Dérivée en x=2: ",(f(x+h)-f(x))/h)
```

Le graphique montre que la pente est négative en $x = -1$ et positive en $x = 2$. Le **signe** de la dérivée indique le sens de la pente, et sa **valeur** montre son intensité. Vérifions avec les [dérivées usuelles](https://example.com) : Pour $f(x) = 2x^2 - 3x + 4$, on a $f'(x) = 4x - 3$. On trouve $f'(-1) = -7$ et $f'(2) = 5$. Les résultats ne sont pas exacts car $h$ n'est pas infiniment petit.


```python
def deriv_f(x):
    return 4*x-3
```

# l’optimisation par descente de gradient


L’**optimisation** vise à minimiser ou maximiser une fonction objectif $f(x)$. Pour trouver le minimum, deux approches sont possibles :

* **Résoudre analytiquement :** $f'(x) = 0$. Cela donne le minimum dans ce cas, mais pas toujours en général (par exemple, en présence de minimums locaux ou de points d'inflexion).
* **Utiliser la descente du gradient :**
  1. On part d’un point initial, par exemple $x_0$.
  2. On calcule la dérivée $f'(x_0)$. Une pente positive ($f'(x_0) > 0$) signifie que si $x$ augmente, $f(x)$ augmente aussi, et inversement.
  3. Pour minimiser $f(x)$, on ajuste $x$ avec un facteur $\alpha$ (*learning rate* ou taux d'apprentissage) :
     $$x_{1} = x_{0} - \alpha f'(x_0)$$
  4. On obtient ainsi une nouvelle valeur $x_1$.
  5. On recalcule $f'(x_1)$ : si la pente est toujours positive, on diminue encore $x$.

La descente du gradient ajuste $x$ en boucle ($x_{t+1} = x_t - \alpha f'(x_t)$) jusqu’à atteindre un minimum, en tenant compte de la pente.


```python
# Descente de gradient
x= 2.0 #valeur aléatoire de x
alpha= 0.01 #pas
iterations = 2450 # nombre d'iterations 

for i in range(iterations):
    grad=deriv_f(x)
    if (grad>0):
        x=x-alpha
    elif (grad<0):
        x=x+alpha
    else:
        print("minimun found YAY, x= ",x)
print("approximate minimun found YAY, x= ",x)
```

On obtient x = 3/4 Avec un pas plus petit (
) et plus d’itérations, on peut affiner le résultat.

# La règle de la chaîne

Avant d'aller plus loin, rappelons une règle mathématique essentielle pour le deep learning : la règle de la chaîne. Elle permet d'entraîner les paramètres des couches cachées d'un réseau. Si $y$ dépend de $u$, qui dépend de $x$, alors :

$$\frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx}$$

Prenons un exemple avec des fonctions dépendantes :
* $u = 2x^2 - x - 2$
* $y = 3u + 1$

$$\frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx} \quad \text{avec} \quad \frac{dy}{du} = 3 \quad \text{et} \quad \frac{du}{dx} = 4x - 1$$

$$\frac{dy}{dx} = 3(4x - 1) = 12x - 3$$

Maintenant, on sait comment $x$ affecte $y$, et on peut appliquer l'algorithme de descente du gradient.


```python
x=2.0
def deriv_y_x(x):
    return 12*x-3
for i in range(iterations):
    grad=deriv_y_x(x)
    if (grad>0):
        x=x-alpha
    elif (grad<0):
        x=x+alpha
    else:
        print("minimun found YAY, x= ",x)
print("approximate minimun found YAY, x= ",x)
```

# Optimisation de plusieurs variables

Jusqu’ici, on a cherché le minimum d’une fonction avec une seule variable x
. Un avantage des méthodes d’optimisation est qu’on peut optimiser plusieurs variables en même temps avec la descente du gradient. Pour cela, il faut calculer la dérivée pour chaque variable.
    

Prenons 3 variables $a$, $b$ et $c$ dans le modèle : $u = 3a^2 - 2a + b^2 + 1$ et $y = 2u + c$. Pour appliquer la descente du gradient, calculons $\frac{dy}{da}$, $\frac{dy}{db}$ et $\frac{dy}{dc}$. Avec la règle de la chaîne :

* Pour $a$ : $\frac{dy}{da} = \frac{dy}{du} \cdot \frac{du}{da} = 2(6a - 2) = 12a - 4$
* Pour $b$ : $\frac{dy}{db} = \frac{dy}{du} \cdot \frac{du}{db} = 2(2b) = 4b$
* Pour $c$ : $\frac{dy}{dc} = 1$

Maintenant, on peut appliquer la descente du gradient.


```python
def deriv_y_a(a):
    return 12*a-4
def deriv_y_b(b):
    return 4*b
def deriv_y_c(c):
    return 1

a=1.0
b=1.0
c=1.0
alpha=0.05
def deriv_y_x(x):
    return 12*x-3
for i in range(iterations):
    grad_a=deriv_y_a(a)
    grad_b=deriv_y_b(b)
    grad_c=deriv_y_c(c)
#variable a
    if (grad_a>0):
        a=a-alpha
    else:
        a=a+alpha
# variable b
    if (grad_b>0):
        b=b-alpha
    else:
        b=b+alpha
#variable c
    if (grad_c>0):
        c=c-alpha
    else:
        c=c+alpha
print("approximate minimum found YAY, a = "+str(a)+" b = "+str(b)+" c= "+str(c))
```

On a trouvé des valeurs qui minimisent y

-c tend vers moins l’infini avec beaucoup d’itérations.

-b vaut 0.

-a vaut 0.3.
